```python
from PIL import Image
import os

def split_image_into_patches(image_path, patch_size, output_folder):
    """
    Splits an image into smaller patches and saves them in the specified output folder.

    :param image_path: Path to the input image.
    :param patch_size: Tuple (patch_width, patch_height) for patch dimensions.
    :param output_folder: Directory to save the patches.
    """
    # Open the image
    image = Image.open(image_path)
    image_width, image_height = image.size
    
    patch_width, patch_height = patch_size
    
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    patch_number = 0
    
    for y in range(0, image_height, patch_height):
        for x in range(0, image_width, patch_width):
            # Define the bounding box for each patch
            box = (x, y, x + patch_width, y + patch_height)
            
            # Crop the patch
            patch = image.crop(box)
            
            # Save the patch
            patch_path = os.path.join(output_folder, f"patch_{patch_number}.png")
            patch.save(patch_path)
            patch_number += 1

    print(f"Saved {patch_number} patches to {output_folder}")

# Example usage
image_path = "horse.png"  # Replace with your image path
output_folder = "output_patches"
patch_size = (32, 32)  # Replace with your desired patch size
split_image_into_patches(image_path, patch_size, output_folder)
```
